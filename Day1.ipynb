# Series (1D - one dimensional labeled array)

# 1. From list

```python
import pandas as pd
s = pd.Series([1,2,3,4,5])
s
```

# custom index

```python
s = pd.Series([1,2,3,4,5],index = ["a","b","c","d","e"])
s
```

```python
s["a"]
```

### From Dict

```python
d = pd.Series(
    {
        "a":100,
        "b":200,
        "c":300   
    }
)
d
```

```python
# series attributes

print("Index:",s.index)
print("Dtype:",s.dtype)
print("Shape:",s.shape)
print("Size:",s.size)
print("values:",s.values)
```

# DataFrame

```python
df
```
