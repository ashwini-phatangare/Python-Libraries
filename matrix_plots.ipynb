# Seaborn - Matrix Plots

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
numeric_data = tips.select_dtypes(include='number')
numeric_data.head()
```

# 1.Heatmap

**Purpose** - Visualizes values in a matrix.commonly used for correlation matrices.

```python
correlation = numeric_data.corr()
sns.heatmap(correlation,annot=True)
plt.title('Correlation Heatmap')
plt.show()
```

# 2. Clustermap

**Purpose** - Combines a heatmap with hierarchical clustering

```python
correlation = numeric_data.corr()
sns.clustermap(correlation,annot=True)
plt.title('Clustermap')
plt.show()
```
