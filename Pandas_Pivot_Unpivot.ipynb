# Pandas Pivot

```python
import pandas as pd
```

# Sample Data for Pivot

```python
df = pd.DataFrame(
    {
        "Name":["A","A","B","B"],
        "Subject":["Math","Science","Math","Science"],
        "Marks":[80,90,70,85]
    }
)
display(df)
```

# Pivot using pivot()

### pivot converts long-format data into wide-format data

```python
pivot_df = df.pivot(
    index = "Name",
    columns = "Subject",
    values = "Marks"
)
display(pivot_df)
```

# Reset The Index

```python
pivot_df = pivot_df.reset_index()
display(pivot_df)
```

# Pivot Table

###### Use ***pivot_table()*** when duplicate combinations  exist and aggregation is required

```python
df_salary = pd.DataFrame({
    "Department":["IT","IT","IT","HR","HR","HR"],
    "Gender":["Male","Female","Male","Male","Female","Female"],
     "Salary":[50000,60000,55000,45000,55000,58000]
})
df_salary
```

```python
salary_pivot = df_salary.pivot_table(
    index = "Department",
    columns = "Gender",
    values = "Salary",
    aggfunc = "mean",
    
)
display(salary_pivot)
```

# Common Aggregation Function

```python
print("SUM")
display(df_salary.pivot_table(
    index = "Department",
    columns = "Gender",
    values = "Salary",
    aggfunc = "sum"
    
))
print("COUNT")
display(df_salary.pivot_table(
    index = "Department",
    columns = "Gender",
    values = "Salary",
    aggfunc = "count"
))
print("MAX")
display(df_salary.pivot_table(
    index = "Department",
    columns = "Gender",
    values = "Salary",
    aggfunc = "max"
))
```

# Unpivot using ***melt()***

#### Unpivot converts wide- format data into long-format data

```python
marks_df = pd.DataFrame(
    {
        "Name":["A","B","C"],
        "Math":[80,70,90],
        "Science":[90,85,95],
        "English":[75,80,88]
    })
marks_df
```

```python
long_df  = marks_df.melt(
    id_vars = "Name",
    var_name = "Subject",
    value_name = "Marks"
)
display(long_df)
    
```

# Unpivot Selected Columns

```python
selected_melt = marks_df.melt(
    id_vars = "Name",
    value_vars =["Math","Science"],
    var_name = "Subject",
    value_name = "Marks"
)
display(selected_melt)
```
