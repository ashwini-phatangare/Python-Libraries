# selecting and filtering Data in Pandas
>**selection** means choosing what data/columns you want
>
>
>**Filtering** means choosing which rows/records you want

# 1. Import Pandas

### Syntax

```python
import pandas as pd
```

# create a Sample DataFrame

```python
data = {
    "Name":["Amit","Sneha","Rahul","Priya","Neha"],
    "Age":[22,23,21,28,24],
    "city":["Pune","Mumbai","hyderabad","Delhi","Mumbai"],
    "Salary":[30000,45000,28000,60000,40000]
}
df = pd.DataFrame(data)
df
```

# Part A - Selecting Single Column

###  Select One column

```python
name_column = df["Name"]
name_column
```

### Select Multiple Columns

```python
selected_columns = df[["Name","Salary"]]
selected_columns
```

# Part B - Selecting Rows with iloc

### Select One Row Using iloc

iloc means **integer-location based selection**

```python
df.iloc[0]
```

# Selects Multiple Rows Using iloc

```python
df.iloc[0:3]
```

### Important

# Select Specific Rows Using iloc

```python
df.iloc[[0,2,4]]
```

# Select Rows and Columns Using iloc

### Syntax
df.iloc[row_selection,column_selection]

```python
df.iloc[0:3,0:2]
```

# Select Specific Rows and columns using iloc

```python
df.iloc[[0,2,4],[0,3]]
```

 # Part C - Selecting Data with loc

 # loc means **labeld based selection**

### Syntax :
df.loc[row_label,column_label]

```python
df.loc[0]
```

# Select Rows Using loc

```python
df.loc[0:2]
```

# Part D :Filtering Data

```python
df[df["Salary"]>40000]
```

# Salary less than 40,000

```python
df[df["Salary"]<40000]
```

# Salary Equal to 40,000

```python
df[df["Salary"]==40000]
```

# Filter Text Values

```python
df[df["city"]=="Pune"]
```

# Part E - Multiple Conditions

# 1. AND condition - &

```python
df[(df["city"]=="Pune") & (df["Salary"]>28000)]
```

# OR Condition - |

```python
df[(df["city"]=="Pune") | (df["city"]=="Mumbai")]
```

# NOT Condition - ~

```python
df[~(df["city"]=="Pune") ]
```

 # Pandas Logical operator

# Filter Multiple Possible Values

```python
df[df["city"].isin(["Pune","Mumbai"])]
```

# Part G - between()

# filter a Numerical Range

```python
df[df["Salary"].between(30000,45000)]
# By default both boundary values are included
```

 # Part H - String Filtering

### str.contains()

```python
df[df["Name"].str.contains("a",case=False,na=False)]
```

# Part I - Filtering missing Values

# Create Dataframe with Misssing Values

```python
df_missing = pd.DataFrame({
    "Name":["Amit","Sneha","Rahul","Priya","Neha"],
    "Age":[22,25,None,28,24],
    "Salary":[30000,None,28000,60000,40000]
}
    
)
df_missing
```

# Find Missing Values with isna()

```python
df_missing[df_missing["Salary"].isna()]
```

# find Non - Missing Values with notna()

```python
df_missing[df_missing["Salary"].notna()]
```
