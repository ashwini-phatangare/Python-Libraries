```python
import numpy as np
```

# argwhere

```python
a = np.array([[1,2,3],[2,3,4],[5,2,1]])
idx = np.argwhere(a==2)
idx
```

# searchsorted

```python
a = np.array([10,20,30,40,50,60])
x = 25
idx = np.searchsorted(a,x)
print(idx)
print(a)
a[2] = 25
print(a)
idx
```

# sort

```python
a =  np.array([4,3,6,2,3,1,7,5,3,2])
sorted = np.sort(a)
sorted
```

```python
a = np.array([3,1,6,4]) # 1,3,4,6 # 0 1 3 2
idx = np.argsort(a)
idx
```

## Unique

**np.unique()** returns the unique (non-repeated) values from an array.



```python
a = np.array([200,300,100,200,100,400,300,500,400,600])
unique_value = np.unique(a)
unique_value
```

# combining array along with axis

```python
a = np.array([[1,2,3],[4,5,6]])
b = np.array([[7,8,9],[10,11,12]])
c = np.concatenate((a,b) ,axis = 0)
c
```

```python
d = np.concatenate((a,b) ,axis = 1)
d
```

# array_split vs split()

```python
a = np.array([1,2,3,4,5])
# split = np.split(a,2) # usedonly for even number of array
array_split = np.array_split(a,3) # it flexible for odd number of array
array_split
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
c = np.array([7,8,9])
column_stack = np.column_stack((a,b,c))
column_stack
```
