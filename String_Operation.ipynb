```python
import pandas as pd
```

# Pandas String Operation

```python
s = pd.Series(["Apple","banana","Cherry","Mango","Apple pie"])
s
```

# Case Conversion

```python
print("Lowercase:")
print(s.str.lower())

print("\nUppercase:")
print(s.str.upper())

print("\nTitle.case:")
print(s.str.title())

print("\nCapitalize:")
print(s.str.capitalize())
```

# Removing Whitespace

```python
names = pd.Series([" John ","Jane","Rahul  ","  Priya "])

print("Original:")
print(names)
print(names.str.len())

print("\nStrip:")
strip=names.str.strip()
print(strip)
print(names.str.len())

print("\nleft strip:")
lstrip=names.str.lstrip()
print(lstrip)
print(names.str.len())

print("\nReft strip:")
rstrip=names.str.rstrip()
print(rstrip)
print(names.str.len())

```

# Searching Strings

```python
print("Contains 'apple':")
print(s.str.contains("apple",case=False))

print("\nStarts with 'A':")
print(s.str.startswith("A"))

print("\nEnds with 'e':")
print(s.str.endswith("e"))
```

# Replace

```python
data = pd.Series(["Python Developer","Python Analyst","Python Trainer"])
print(data.str.replace("Python","Data",regex=False))
```

# Split Strings

```python
names = pd.Series(["John Doe","Jane Smith","Rahul  Patil"])

print("Split:")
print(names.str.split(" "))

print("\nSplit into columns:")
print(names.str.split(" ",expand=True))
      
```

# String Slicing

```python
codes = pd.Series(["IND101","IND202","USA303","UK404"])

print("First 3 Characters:")
print(codes.str.slice(0,3))

print("Last 3 Characters:")
print(codes.str.slice(-3))

print("\nFirst Character:")
print(codes.str[0])
```

# Extract Using Regular Expression

```python
codes = pd.Series(["A-101","B-202","C-303"])

print("Extract numbers:")
print(codes.str.extract(r"(\d+)",expand=False))
print("\nExtract alphabet and number:")
print(codes.str.extract(r"([A-Z])-(\d+)",expand=True)
      
      
      
```

# Count Occurrences

```python
fruits = pd.Series(["banana","apple","mango"])
print(fruits.str.count("a"))
```

# Find All Matches

```python
text = pd.Series(["Python 123","SQL 456","POWERBI 789"])
text.str.findall(r"\d+")
```

# Join

```python
words = pd.Series(["ABC","DATA","SQL"])
words.str.join("-")
```
