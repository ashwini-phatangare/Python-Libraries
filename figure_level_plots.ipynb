# Seaborn - Figure - Level - Plots

This notebook covers Pair Plot,Joint Plot,FaceGrid

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
tips.head()
```

# 1.Pair Plot

**Purpose** - Shows Pairwise relationships between multiple numerical variables

```python
sns.pairplot(tips,hue='sex')
plt.show()
```

# 2.Joint Plot

**Purpose** - shows the relationship between two variables together with their individual distributions

```python
sns.jointplot(data=tips,x='total_bill',y='tip')
plt.show()
```

# 3.FacetGrid

**Purpose** - Creates multiple plots for different subsets of the data

```python
g =  sns.FacetGrid(tips,col='sex')
g.map_dataframe(sns.scatterplot,x='total_bill',y='tip')
plt.show()
```

# 4. PairGrid

**Purpose** - Gives more control over customized pairwise plots

```python
g = sns.PairGrid(tips,vars=['total_bill','tip','size'])
g.map(sns.scatterplot)
plt.show()
```

# JointGrid

**purpose** - Provides a customizable framework for a joint plot

```python
g = sns.JointGrid(data=tips,x='total_bill',y='tip')
g.plot(sns.scatterplot,sns.histplot)
plt.show()
```

# 6.Displot

**Purpose** - Figure - Level distribution visualization

```python
sns.displot(data=tips,x='total_bill',col='sex',kde=True)
plt.show()
```

# 7. LM Plot

**purpose** - Figure - Level regression visualization with grouping.

```python
sns.lmplot(data=tips,x='total_bill',y='tip',hue='sex',col='time')
plt.show()
```
