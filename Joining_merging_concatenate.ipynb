# merging,joining and concatenate

```python
import numpy as np
import pandas as pd
```

```python
df1 = pd.DataFrame(
    {
        "ID":[1,2,3],
        "Name":["Alice","Bob","Charlie"],
        "Dept":["HR","IT","Sales"]
    }
)
df1
   
```

```python
df2 = pd.DataFrame(
    {
        "ID":[1,2,4],
        "Salary":[50000,60000,45000]
    }
)
df2
```

```python
pd.merge(df1,df2,how="inner",on="ID")
```

```python
pd.merge(df1,df2,how="left",on="ID")
```

```python
pd.merge(df1,df2,how="right",on="ID")
```

```python
pd.merge(df1,df2,how="outer",on = "ID")
```

# Joining two DataFrames

```python
sales_Data = pd.DataFrame(
    {
        "ID":[101,102,103,104],
        "CustID":["C001","C002","C003","C004"],
        "Amount":[1200,900,1500,700]
    }
)
```

```python
sales_Data
```

```python
cust_data = pd.DataFrame(
    {
        "CustID":["C001","C002","C004"],
        "Name":["Alice","Bob","David"],
        "City":["Delhi","Mumbai","Pune"]
    }
)
```

```python
cust_data
```

```python
sales_df_indexed = sales_Data.set_index('CustID')
customer_df_indexed = cust_data.set_index('CustID')
```

```python
joined_df = sales_df_indexed.join(customer_df_indexed)
print(joined_df)
```

# concatenate

```python
table1 = pd.DataFrame(
    {
        "ID":[1,2,3],
        "A":["A1","A2","A3"],
        "B":["B1","B2","B3"]
    })
table1
```

```python
table2 = pd.DataFrame(
    {
        "ID":[1,2,3],
        "A":["A1","A2","A3"],
        "B":["B1","B2","B3"],
        "C":["C1","C2","C3"]
    })
table2
```

```python
pd.concat([table1,table2],axis=0) # row wise
```

```python
pd.concat([table1,table2],axis=1) # column wise
```
