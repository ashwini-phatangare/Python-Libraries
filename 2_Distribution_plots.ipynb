```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load DataSet

tips = sns.load_dataset('tips')
tips.head()
```

# 1. Histogram

**purpose** - Shows how frequently values occur in ranges

```python
sns.histplot(data=tips,x='total_bill')
plt.title('Histogram - Total Bill')
plt.show()
```

# KDE Plot - Kernel Density Estimate

**Pupose** : Shows a smooth estimate of the distribution.

```python
sns.kdeplot(data=tips,x='total_bill')
plt.title('KDE plot - Total Bill')
plt.show()
```

# 3.ECDF Plot 

**purpose** - shows the cumulative proportion of observation

```python
sns.ecdfplot(data=tips,x='total_bill')
plt.title('ECDE Plot - Total Bill')
plt.show()
```

# 4. Rug Plot

**Purpose** - Shows Individual observations as small marks along an axis

```python
sns.rugplot(data=tips,x='total_bill')
plt.title('Rug Plot - Total Bill')
plt.show()
```

# 5. Displot

**purpose** - A figure- level function for distribution visualization

```python
sns.displot(data=tips,x='total_bill',kde=True)
plt.show()
```
