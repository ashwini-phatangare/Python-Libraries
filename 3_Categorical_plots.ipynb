# Seaborn - Categorical Plots

This notebook covers Bar,Count,Box,Violin,Strip,Swarm,and Points plots

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load DataSet

tips = sns.load_dataset('tips')
tips.head()
```

# 1.Bar Plot

**Pupose** - compares an aggregate value across categories

```python
sns.barplot(data=tips,x='day',y='total_bill')
plt.title('average bill by day')
plt.show()
```

# 2.Count Plot 

**Purpose** - Counts how many observations belong to each category

```python
sns.countplot(data=tips,x='day')
plt.title('Number of records by Day')
plt.show()
```

# 3.Box Plot

**Purpose** - Shows median,spread,quartiles and possible outliers

```python
sns.boxplot(data=tips,x='day',y='total_bill')
plt.title("Bill Distribution by Day")
plt.show()
```

# 4.Violin Plot

**Purpose** - Combines distribution shape with summary statistics

```python
sns.violinplot(data=tips,x='day',y='total_bill')
plt.title("Distribution By Day")
plt.show()
```

# 5.Strip Plot

**Purpose** - Displays Individual observations for each category

```python
sns.stripplot(data=tips,x='day',y='total_bill')
plt.title("Individual bills by day")
plt.show()
```

# 6.Swarm Plot

```python
sns.swarmplot(data=tips,x='day',y='total_bill')
plt.title("Indivaul Bills By day")
plt.show()
```

# 7.Point Plot

**Purpose** - Compares an estimated central value across categories

```python
sns.pointplot(data=tips,x='day',y='total_bill')
plt.title('Point Plot')
plt.show()
```
