# Matplotlib  Graphs for Data Analytics

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

# 1.Line Plot

```python
months = ["Jan","Feb","Mar","Apr","May"]
sales = [100,120,150,130,180]

plt.plot(months,sales)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

# Customize Line Plot

```python
plt.plot(
    months,
    sales,
    marker ="*",
    linestyle ="--",
    linewidth =2,
    label ="Sales"
)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.grid()
plt.legend()

plt.show()

    
```

# 2. Bar Chart

```python
products = ["Laptop","Mobile","Tablet","Watch"]
sales = [80,120,60,40]

plt.bar(products,sales , color = "yellow")

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Sales")

plt.show()
```

# Customize Bar Graphs

```python
plt.bar(
    products,
    sales,
    width=0.6,
    edgecolor="black"
)

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Units Sold")

plt.grid(axis = "y")
plt.show()
```

# 3.Horizontal Bar Graph

```python
departments = [
    "Human Resources",
    "Finance",
    "Data Analytics",
    "Marketing"
]

employees = [15,20,35,25]

plt.barh(departments,employees)

plt.title("Employees by Department")
plt.xlabel("Employee")
plt.ylabel("Department")
plt.show()
```

# customize of Horizontal Bar Graph

```python
plt.barh(
    departments,
    employees,
    height = 0.6,
    edgecolor = "black",
    color = ["red","blue","yellow","violet"]
)

plt.grid(axis = "x")
plt.show()
```

# 4.Scatter Plot

```python
experience = [1,2,3,4,5,6,7,8]
salary = [25,28,32,38,42,48,55,62]

plt.scatter(experience,salary)

plt.title("Experience vs Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")

plt.show()
```

# customization of Scatter Plot

```python
plt.scatter(
    experience,
    salary,
    s=100,
    alpha = 0.7,
    marker = "o",
    color = "yellow"
)

plt.grid()
plt.show()
```

# 5 .Pie Chart

```python
categories = ["Food","Travel","Rent","Shopping"]
expenses = [25,15,40,20]

plt.pie(
    expenses,
    labels=categories,
    autopct="%1.1f%%"
)

plt.title("Monthly Expenses")
plt.show()
```

# customization Pie Chart

```python
plt.pie(
    expenses,
    labels=categories,
    autopct="%1.1f%%",
    startangle=90,
    explode=[0,0,0.1,0]
)

plt.title("Monthly Expenses")
plt.show()
```

# Histogram 

```python
marks = [
    45,52,55,60,61,
    63,65,67,68,70,
    72,73,75,76,78,
    80,82,85,88,90
]
plt.hist(
    marks,
    bins =5,
    edgecolor ="black",
    color = "yellow"
)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
salary = [
    25,28,30,31,
    32,35,36,38,
    40,42,45,100
]
plt.boxplot(salary)

plt.title("Salary Distribution")
plt.ylabel("Salary")

plt.show()
```

```python
plt.boxplot(
    salary,
    vert=False,
    showmeans=True
)
plt.title("Salary Distribution")
plt.show()
```

# Area Plot

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
months = ["Jan","Feb","Mar","Apr","May"]
sales = [100,120,150,130,180]

plt.plot(months,sales)
plt.fill_between(months,sales,alpha=0.3)

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

# Stack Plot

```python
months = ["Jan","Feb","Mar","Apr","May"]

online = [20,30,40,45,55]
store = [30,35,40,42,48]
partner = [10,15,20,18,22]

plt.stackplot(
    months,
    online,
    store,
    partner,
    labels=["online","store","partner"]
)

plt.title("Stack Channel")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()

plt.show()
```

# ViolinPlot

```python
group_a = [45,50,52,55,58,60,61,63,65]

group_b = [50,55,60,62,65,68,70,72,80]

plt.violinplot(
    [group_a,group_b],
    showmeans=True
)
plt.xticks(
    [1,2],
    ["Group A","Group B"]
)

plt.title("Marks Distribution")
plt.ylabel("Marks")

plt.show()
```

```python
products = ["Laptop","Mobile","Tablet","Watch"]
sales = [80,120,60,40]

plt.subplot(2,2,1)
plt.bar(products,sales,color = "purple",edgecolor="black")
plt.title("Bar Graph")

plt.subplot(2,2,2)
plt.scatter(products,sales,color = "black",marker="^")
plt.title("Scatter Plot")


plt.subplot(2,2,3)
plt.hist(sales,bins=3,color = "yellow",edgecolor="purple")
plt.title("Histogram")

plt.subplot(2,2,4)
plt.plo(products,sales,color = "blue")
plt.title("Area Plot")

plt.tight_layout()
plt.savefig("subplot.png")
plt.show()
```
