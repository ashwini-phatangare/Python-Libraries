# Standard Deviation and Variance

**Variance**=
Variance measures how far the data values are spread out from the mean.

Simple Marathi मध्ये:
Mean पासून प्रत्येक value किती दूर आहे आणि data किती spread आहे हे Variance सांगते.

**Standard Deviation** =
Standard deviation measures the amount of variation or spread of data values around the mean.

Simple Marathi मध्ये:
Data values mean च्या आसपास किती पसरलेल्या आहेत हे Standard Deviation सांगते.

Variance चा square root = Standard Deviation    

```python
marks = np.array([80,85,90,95,100])
marks
```

```python
var = np.var(marks)
var
```

```python
std = np.std(marks)
std
```

```python
m = np.array([50,70,90,110,130])
m
```

```python
print("Mean:",np.mean(m))
print("Variance:",np.var(m))
print("Standard Deviation:",np.std(m))
```

```python
# Conclusion :
# variance == high
# std dev = high
# students marks are very different from each other
```

# argmin and argmax

**argmin()** returns the index of the minimum value in an array.

**argmax()** returns the index of the maximum value in an array.

```python
arr = np.array([20,10,50,30])
print("min value:",np.min(arr))
print("max value:",np.max(arr))
```

```python
print("argmin:",np.argmin(arr))
print("argmax:",np.argmax(arr))
```

```python
a = np.arange(1,13).reshape(3,4)
a
```

```python
# overall array
print("min value:",np.min(a))
print("max value:",np.max(a))
print("argmin:",np.argmin(a))
print("argmax:",np.argmax(a))
```

```python
# row wise
print("min value:",np.min(a,axis =1))
print("max value:",np.max(a,axis = 1))
print("argmin:",np.argmin(a,axis = 1))
print("argmax:",np.argmax(a,axis=1))
```

```python
# column wise
print("min value:",np.min(a,axis=0))
print("max value:",np.max(a,axis=0))
print("argmin:",np.argmin(a,axis =0))
print("argmax:",np.argmax(a,axis=0))
```

# np.where

**np.where()** returns the indices where a given condition is True.


Array मधल्या कोणत्या values आपली condition satisfy करतात, त्यांचा index शोधण्यासाठी where() वापरतो.

```python
arr = np.array([20,12,23,20,30,20,20,23])
idx = np.where(arr == 20)
idx
```

```python
marks = np.array([35,60,25,80,45])
result = np.where(marks >= 35,"Pass","Fail")
result
```
